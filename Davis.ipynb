```python
import pandas as pd
import numpy as np

df = pd.read_csv("Instagram_Analytics.csv")

print("Dataset Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percent
})

missing_table = missing_table[missing_table["Missing Values"] > 0]

print(missing_table)
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
df["post_datetime"] = pd.to_datetime(
    df["post_datetime"],
    errors="coerce"
)

df["post_date"] = pd.to_datetime(
    df["post_date"],
    errors="coerce"
)
```

```python
print(df[["post_datetime", "post_date"]].isnull().sum())
```

```python
categorical_columns = [
    "account_type",
    "media_type",
    "content_category",
    "traffic_source",
    "day_of_week",
    "performance_bucket_label"
]

for col in categorical_columns:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
```

```python
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
numerical_columns = [
    "follower_count",
    "post_hour",
    "likes",
    "comments",
    "shares",
    "saves",
    "reach",
    "impressions",
    "engagement_rate",
    "followers_gained",
    "caption_length",
    "hashtags_count"
]
```

```python
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
non_negative_columns = [
    "follower_count",
    "post_hour",
    "likes",
    "comments",
    "shares",
    "saves",
    "reach",
    "impressions",
    "followers_gained",
    "caption_length",
    "hashtags_count"
]

for col in non_negative_columns:
    print(col, (df[col] < 0).sum())
```

```python
for col in non_negative_columns:
    df.loc[df[col] < 0, col] = np.nan
```

```python
for col in non_negative_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
print(df["engagement_rate"].describe())
```

```python
Q1 = df["engagement_rate"].quantile(0.25)
Q3 = df["engagement_rate"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["engagement_rate"] < lower_bound) |
    (df["engagement_rate"] > upper_bound)
]

print("Engagement Rate Outliers:", len(outliers))
```

```python
print(
    "Rows where impressions < reach:",
    (df["impressions"] < df["reach"]).sum()
)
```

```python
print(
    "Rows with zero/negative reach:",
    (df["reach"] <= 0).sum()
)
```

```python
df["calculated_engagement_rate"] = (
    (df["likes"] + df["comments"] + df["shares"] + df["saves"])
    / df["reach"]
) * 100
```

```python
df["calculated_engagement_rate"] = np.where(
    df["reach"] > 0,
    (
        (df["likes"] + df["comments"] + df["shares"] + df["saves"])
        / df["reach"]
    ) * 100,
    0
)
```

```python
df["total_engagement"] = (
    df["likes"]
    + df["comments"]
    + df["shares"]
    + df["saves"]
)
```

```python
df["follower_growth_rate"] = np.where(
    df["follower_count"] > 0,
    (df["followers_gained"] / df["follower_count"]) * 100,
    0
)
```

```python
df["impression_reach_ratio"] = np.where(
    df["reach"] > 0,
    df["impressions"] / df["reach"],
    0
)
```

```python
df = df.drop(columns=["engagement_rate"])
```

```python
df = df.rename(
    columns={
        "calculated_engagement_rate": "engagement_rate"
    }
)
```

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv(
    "Instagram_Analytics_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
